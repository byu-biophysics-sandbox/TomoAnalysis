```python
import os
import numpy as np
import SimpleITK as sitk
from PIL import Image

def convert_mha_to_png(input_dir, output_dir):
    # Ensure the global output directory exists
    os.makedirs(output_dir, exist_ok=True)
    
    # Iterate over all files in the input directory
    for root, dirs, files in os.walk(input_dir):
        for file in files:
            if file.endswith('.mha'):
                # Construct the full file path
                file_path = os.path.join(root, file)
                
                # Read the .mha file
                image = sitk.ReadImage(file_path)
                image_array = sitk.GetArrayFromImage(image)
                
                # Determine the shortest axis
                shortest_axis = np.argmin(image_array.shape)
                
                # Create output directory for this file
                base_name = os.path.splitext(file)[0]
                folder_name = f"{os.path.basename(root)}_{base_name[-4:]}"
                file_output_dir = os.path.join(output_dir, folder_name)
                os.makedirs(file_output_dir, exist_ok=True)
                
                # Iterate over slices along the shortest axis
                for i in range(image_array.shape[shortest_axis]):
                    if shortest_axis == 0:
                        slice_array = image_array[i, :, :]
                    elif shortest_axis == 1:
                        slice_array = image_array[:, i, :]
                    else:
                        slice_array = image_array[:, :, i]
                    
                    # Normalize the slice to 0-255
                    slice_array = ((slice_array - np.min(slice_array)) / (np.max(slice_array) - np.min(slice_array)) * 255).astype(np.uint8)
                    
                    # Convert to image and save
                    slice_image = Image.fromarray(slice_array)
                    slice_image.save(os.path.join(file_output_dir, f"{i:05d}.png"))

# Example usage
convert_mha_to_png('/home/matiasgp/groups/fslg_imagseg/nobackup/archive/segmentation_data/3d_segmentations/segmentations', "/home/matiasgp/groups/fslg_imagseg/nobackup/archive/segmentation_data/3d_segmentations/annotations")
```
